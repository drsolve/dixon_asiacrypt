```python
# Dixon Resultant Matrix Size Bounds Comparison
# Data generation for Appendix C, Table 6

def fuss_catalan(n, d):
    """
    Compute the Fuss-Catalan number C_n^(d)
    Formula: C_n^(d) = 1/(n(d-1)+1) * binomial(nd, n)
    This is the Dixon matrix size bound from Corollary 1(2)
    """
    numerator = binomial(n*d, n)
    denominator = n*(d-1) + 1
    return numerator // denominator

def unrestricted_bound(n, d):
    """
    Compute the unrestricted bound from Corollary 1(1)
    This counts all lattice paths without boundary constraints
    Formula: binomial(n(d-1), n-1)
    """
    total_height = (n-1) * (d-1)
    num_steps = n-1
    return binomial(total_height + num_steps, num_steps)

def bound_from_paper_34(n, d):
    """
    Compute the bound from [34] - equation (5)
    Formula: (d*n)^(n-1) / n!
    This uses Newton polytope volume estimates for multi-homogeneous systems
    """
    numerator = (d * n)^(n-1)
    denominator = factorial(n)
    return floor(numerator / denominator)

def variable_bound(n, d):
    """
    Compute the variable-based trivial bound from equation (4)
    Formula: b0 = (n-1)! * product(m_i) where m_i = deg_xi(Res(F))
    """

    return factorial(n-1) * d^((n-1))

# Generate table data
print("Dixon Matrix Size Bounds Comparison")
print("=" * 90)
print(f"{'n':<4} {'d':<4} {'FC Bound':<15} {'Unrestricted':<15} {'Bound [34]':<15} {'Variable Bound':<15}")
print("-" * 90)

# Test configurations from Table 7
test_configs = [
    (3, 2), (3, 3), (3, 4), (3, 5),
    (4, 2), (4, 3), (4, 4), (4, 5),
    (5, 2), (5, 3), (5, 4), (5, 5),
    (6, 2), (6, 3), (6, 4), (6, 5),
    (7, 2), (7, 3), (7, 4),
    (8, 2), (8, 3),
    (9, 2),
    (10, 2),    (4, 65536)

]

for n, d in test_configs:
    fc = fuss_catalan(n, d)
    unrest = unrestricted_bound(n, d)
    b34 = bound_from_paper_34(n, d)
    varbound = variable_bound(n, d)
    
    print(f"{n:<4} {d:<4} {fc:<15} {unrest:<15} {b34:<15} {varbound:<15}")

print("=" * 90)
print("\nNotes:")
print("- FC Bound: Fuss-Catalan bound from Corollary 1(2), achieved for generic systems")
print("- Unrestricted: Bound from Corollary 1(3) without boundary constraints")  
print("- Bound [34]: Newton polytope volume estimate from Kapur et al. [34]")
print("- Variable Bound: Trivial variable-based bound from equation (4)")
print("  (Variable bound formula is approximate for some cases)")
```
