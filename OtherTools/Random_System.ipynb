```python
### from sage.all import *
import itertools


def random_polynomial_over_finite_field(n, d, p, m, r):
    F = GF(p, names='t')
    R = PolynomialRing(F, n, 'x', order='lex')
    x = R.gens()
    ps = []
    all_exponents = list(itertools.product(range(d + 1), repeat=n))
    valid_exponents = [exp for exp in all_exponents if sum(exp) <= d] #all_exponents#
    r0 = r
    if r > binomial(n+d,d) or r == 0:
        r0 = binomial(n+d,d)
    for i in range(m):
        poly = 0
        used_indices = set()
        while len(used_indices) < r0:
            idx = randint(0, len(valid_exponents) - 1)
            if idx not in used_indices:
                exponents = valid_exponents[idx]
                coefficient = F.random_element()
                term = coefficient * prod([x[j] ** exponents[j] for j in range(n)])
                poly += term
                used_indices.add(idx)
        ps.append(poly)
    return ps


# Example usage
r = 0   # monomial number
n = 6   # number of variables
d = 2   # degree of polynomial
p = 65537  # characteristic of the finite field
m = 6   # number of equations

ps = random_polynomial_over_finite_field(n, d, p, m, r)
ps
```
